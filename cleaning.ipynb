# Cleaning check_dates csv for getting high count sitters

```python
import pandas as pd
import re
```

```python
#import check_dates csv
check_dates_df = pd.read_csv('check_dates.csv')
```

```python
#remove sitters with "unidentified" in Sitter column
check_dates_df = check_dates_df[~check_dates_df['Sitter'].str.contains('unidentified', flags=re.IGNORECASE, na=False)]
# also, remove (Multiple Portraits)     
check_dates_df = check_dates_df[~check_dates_df['Sitter'].str.contains('\(Multiple Portraits\)', flags=re.IGNORECASE, na=False)]    

#get high count sitters
sitter_counts = check_dates_df['Sitter'].value_counts()
#removing rows with mediaURL empty
check_dates_df = check_dates_df[check_dates_df['mediaURLs'].notna()]

```

```python
pd.set_option('display.max_rows', None)
sitter_counts
```

```python
#keep sitters with more than 3 entries
high_count_sitters = sitter_counts[sitter_counts > 3].index.tolist()
high_count_sitters_df = check_dates_df[check_dates_df['Sitter'].isin(high_count_sitters)]
```

```python
high_count_sitters_df = high_count_sitters_df.reset_index(drop=True)
high_count_sitters_df.head()
#drop Unnamed columns
```

```python
#export
high_count_sitters_df.to_csv('high_count_sitters.csv', index=False)
```
